```matlab
% Load NIfTI file
filename = '/rds/general/user/ka325/home/synthseg/SynthSeg/T1_parc.nii.gz';
img = niftiread(filename);

% Get dimensions
[rows, cols, slices] = size(img);

% Create figure
f = figure;
colormap gray;

% Show first slice
slice_num = 83;
hImg = imagesc(img(:,:,slice_num));
axis image off;
title(['Slice: ', num2str(slice_num)]);

% Create slider
s = uicontrol('Style','slider',...
    'Min',1,'Max',slices,'Value',1,...
    'SliderStep',[1/(slices-1) 10/(slices-1)],...
    'Position',[150 20 300 20]);

% Listener (this makes it WORK in Jupyter)
addlistener(s,'Value','PostSet',@(src,event) updateSlice(round(s.Value), img, hImg));

% Update function
function updateSlice(slice_num, img, hImg)
    set(hImg, 'CData', img(:,:,slice_num));
    title(['Slice: ', num2str(slice_num)]);
end
```
